```python
# ============================================================
# STUDENT-FRIENDLY WORLD HAPPINESS DASHBOARD
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

# ------------------------------------------------------------
# STEP 1: Upload the CSV file
# ------------------------------------------------------------
from google.colab import files
print("Please upload your CSV file...")
uploaded = files.upload()

# ------------------------------------------------------------
# STEP 2: Load the dataset
# ------------------------------------------------------------
df = pd.read_csv("world_happiness_dataset (2).csv")
print("Dataset loaded successfully!")
print(df.head())   # Show first few rows

# ------------------------------------------------------------
# STEP 3: Find the Top 3 Happiest Countries
# ------------------------------------------------------------
top3 = df.nlargest(3, "Happiness_Score")
print("\nTop 3 Happiest Countries:")
print(top3[["Country", "Happiness_Score"]])

# ------------------------------------------------------------
# STEP 4: Find the Country with the Lowest Happiness Score
# ------------------------------------------------------------
lowest = df.nsmallest(1, "Happiness_Score").iloc[0]
lowest_country = lowest["Country"]
lowest_freedom = lowest["Freedom_to_Make_Choices"]

print("\nLeast Happy Country:", lowest_country)
print("Freedom Score:", lowest_freedom)

# ------------------------------------------------------------
# STEP 5: Matplotlib Chart (Top 3 Happiest)
# ------------------------------------------------------------
plt.figure(figsize=(10,6))
sns.barplot(data=top3, x="Country", y="Happiness_Score", palette="coolwarm")
plt.title("Top 3 Happiest Countries (Matplotlib)")
plt.xlabel("Country")
plt.ylabel("Happiness Score")
plt.show()

# ------------------------------------------------------------
# STEP 6: Plotly Chart (Top 3 Happiest)
# ------------------------------------------------------------
fig = px.bar(
    top3,
    x="Country",
    y="Happiness_Score",
    color="Country",
    title="Top 3 Happiest Countries (Interactive Plotly Chart)",
    text="Happiness_Score"
)
fig.update_traces(textposition="outside")
fig.show()

# ------------------------------------------------------------
# STEP 7: Matplotlib Chart (Freedom Score of Least Happy Country)
# ------------------------------------------------------------
plt.figure(figsize=(6,4))
plt.bar([lowest_country], [lowest_freedom], color="red")
plt.title(f"Freedom Score for {lowest_country}")
plt.ylabel("Freedom to Make Choices")
plt.show()

# ------------------------------------------------------------
# STEP 8: Plotly Chart (Freedom Score of Least Happy Country)
# ------------------------------------------------------------
fig2 = px.bar(
    x=[lowest_country],
    y=[lowest_freedom],
    labels={"x": "Country", "y": "Freedom Score"},
    title=f"Freedom Score for {lowest_country} (Interactive Plotly)"
)
fig2.show()
```
